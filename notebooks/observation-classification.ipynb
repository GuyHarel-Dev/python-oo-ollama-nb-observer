```python
import json
from pathlib import Path
import model_observability.core.log_file_utils as log_file_utils

# produce responses
#   ython -m model_observability.chat_observer "C:\GH-ORDI\PROJ\AI\Model-Observability\observations\classify-temp-(0.0-0.8)-7x30.json"

file_root = r"C:\GH-ORDI\DATA\Data-Model-Observability\LLM-Model-Observability\classify_statement_v1-temp-(0.0-0.8)-7x30"

file_folder = r""

folder_path = Path(file_root + file_folder)
log_files_results = log_file_utils.load_log_files_results(folder_path)
```

```python
# display options and response content
from dataclasses import dataclass

@dataclass
class OptionContent:
    option_name: str
    option_value : float 
    content_value : str

options_contents : list[OptionContent] = []
for log_file_name, chat_results in log_files_results.items():
    for result in chat_results: 
        option_content = OptionContent(option_name="temp", option_value=result.chat_options.temperature, content_value=str(result.chat_response.content).strip())
        options_contents.append(option_content)

options_contents

```

```python
# get all the temperature, content pairs
from pandas import Series

temperature_values = [content.option_value for content in options_contents]
temperature_values = Series(temperature_values).unique()

@dataclass
class TempResponse:
    temperature: float
    response: str
    inference_tokens: int = 0

temp_responses : list[TempResponse] = []

for log_file_name, chat_results in log_files_results.items():
    for result in chat_results:
        temp_responses.append(TempResponse(temperature=result.chat_options.temperature, response=str(result.chat_response.content).strip(), inference_tokens=int(result.chat_response.inference_tokens)))

temp_responses 
```

```python
# for each temperature get all responses content by value

#negative_response = "POSITIVE"
#positive_response = "NEGATIVE"
#neutral_response = "NEUTRAL"

temps = Series([temp_resp.temperature for temp_resp in temp_responses]).unique()

for temperature in temps:
    positive_response = len([temp_resp.temperature for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="POSITIVE"])
    negative_response = len([temp_resp.temperature for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="NEGATIVE"])
    neutral_response = len([temp_resp.temperature for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="NEUTRAL"])

    data = {"temperature": temperature, "positive": positive_response, "negative": negative_response, "neutral": neutral_response}

```

```python
# for each temperature get all reference tokens by responses content

temps = Series([temp_resp.temperature for temp_resp in temp_responses]).unique()

# temp and list of 
for temperature in temps:
    positive_tokens = [temp_resp.inference_tokens for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="POSITIVE"]
    negative_tokens = [temp_resp.inference_tokens for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="NEGATIVE"]
    neutral_tokens = [temp_resp.inference_tokens for temp_resp in temp_responses if temp_resp.temperature == temperature and temp_resp.response =="NEUTRAL"]

    data = {"temperature": temperature, "positive_tokens": positive_tokens, "negative_tokens": negative_tokens, "neutral_tokens": neutral_tokens}
    print(data)
```
